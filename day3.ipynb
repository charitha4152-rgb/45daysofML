# NumPy

## One dimensional array

```python
import numpy as np
arr=np.array([10,20,25,40,60])

print(arr)
```

```python
import numpy as np
a = [1, 2, 3, 4]
arr = np.array(a)
print("List in python : ", a)
print("Numpy Array in python :",
    arr)
```

## Two dimensional array

```python
import numpy as np
arr=np.array([
    [1,2,3],
    [7,5,9],
    [4,3,6]
])

print(arr)
```

```python
import numpy as np
list_1 = [1, 2, 3, 4]
list_2 = [5, 6, 7, 8]
list_3 = [9, 10, 11, 12]
sample_array = np.array([list_1,
                        list_2,
                        list_3])
print("Numpy multi dimensional array in python\n",
    sample_array)
```

## Shape

```python
print("Shape of the array : ", sample_array.shape)
```
