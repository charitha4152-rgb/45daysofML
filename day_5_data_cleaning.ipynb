# Data cleaning

## creating a dataset

```python
import pandas as pd
import numpy as np
data = {
    "Name":["Alice", "Bob", "Charlie", "David", None],
    "Age":[25, np.nan, 35, 40, 22],
    "Marks":[85, 60, np.nan, 90, 75],
    "City":["New York", "Los Angeles", None, "Chicago", "Boston"]
}

df = pd.DataFrame(data)
print(df)
```

```python
print(df.isnull().sum())  #checking missing values
```

```python
df_cleaned = df.dropna()  #dropping rows with missing values
print(df_cleaned)
```

```python
print(df.duplicated().sum())  #checking duplicates
```

```python
df = df.drop_duplicates() #dropping duplicates
```

```python
print(df.info())
print(df.describe())
```

```python
print(df["Marks"].mean())
print(df["Age"].mean())
```
