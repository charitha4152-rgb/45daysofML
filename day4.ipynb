### Dataframe(2D)

```python
import pandas as pd
data={
    "Name":['Alice','Bob','Charlie'],
    "Age":[25,30,35],
    "City":['New York','Los Angeles','Chicago']
}
df=pd.DataFrame(data)
print(df)
```
